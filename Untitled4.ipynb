```python
# --- EJERCICIO: ENVÍO Y NIVEL DE FIDELIDAD ---

total = 850          # ← cambia
es_miembro = False   # ← True o False
compras = 7          # ← cambia

# 1. Envío: más de 1000 → gratis;
#    miembro → gratis; si no → $99
if total > 1000 or es_miembro:
    envio = 0
else:
    envio = 99
print(f"Envío: ${envio}")

# 2. Nivel: 10 o más → Oro;
#    de 5 a 9 → Plata; menos → Bronce
if compras >= 10:
    nivel = "Oro"
elif compras >= 5:
    nivel = "Plata"
else:
    nivel = "Bronce"
print(f"Nivel: {nivel}")
```

```python
# --- EJERCICIO: RECORRIDO DE PRECIOS ---

# ← cambia por los precios de tu compra
precios = [120, 45, 300, 80]

# 1. Imprime cada precio con un for
for precio in precios:
    print(precio)

# 2. Suma el total
total = 0
for precio in precios:
    total += precio
print("Total:", total)

# 3. ¿Cuántos cuestan más de 100?
caros = 0
for precio in precios:
    if precio > 100:
        caros += 1
print("Más de 100:", caros)
```

```python
# --- EJERCICIO: SIMULADOR DE AHORRO ---

ahorro = 0       # ← lo que ya tienes
meta = 5000      # ← cambia
por_mes = 800    # ← cambia
meses = 0

# Mientras no llegues a la meta:
# suma lo del mes y cuenta un mes más
while ahorro < meta:
    ahorro += por_mes
    meses += 1

# ¿Cuántos meses tardaste?
print("Meses necesarios:", meses)
```

```python
# --- EJERCICIO: CUENTA, PROPINA Y DIVISIÓN ---

cuenta = 1200    # ← cambia
propina = 10     # ← cambia (%)
amigos = 4       # ← cambia

# 1. Regresa la cuenta más la propina
def con_propina(cuenta, porcentaje):
    return cuenta + (cuenta * porcentaje / 100)

# 2. Regresa cuánto paga cada persona
def por_persona(total, personas):
    return total / personas

# 3. Úsalas con tus datos e imprime
total_cuenta = con_propina(cuenta, propina)
cada_uno = por_persona(total_cuenta, amigos)

print("Total con propina:", total_cuenta)
print("Por persona:", cada_uno)
```
